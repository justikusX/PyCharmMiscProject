```python
import numpy as np
```

```python
# 1. Массив из 50 случайных чисел и гистограмма с 5 бинами
arr1 = np.random.randint(1, 101, 50)

hist, bins = np.histogram(arr1, bins=5)

print("1. Массив:")
print(arr1)
print("Гистограмма:", hist)
print("Границы бинов:", bins)
```

```python
# 2. Матрица 3x4, заполненная единицами
matrix2 = np.ones((3, 4))

print("\n2. Матрица 3x4:")
print(matrix2)
```

```python

# 3. Массив из 10 элементов с равномерным шагом от -1 до 1
arr3 = np.linspace(-1, 1, 10)

print("\n3. Массив:")
print(arr3)
```

```python
# 4. Ковариация между двумя массивами
a = np.array([1, 2, 3, 4, 5])
b = np.array([2, 4, 5, 4, 6])

covariance = np.cov(a, b)

print("\n4. Матрица ковариации:")
print(covariance)
print("Ковариация между массивами:", covariance[0, 1])
```

```python
# 5. Матрица 4x4 и её псевдообратная матрица
matrix5 = np.random.randint(1, 10, (4, 4))

pseudo_inverse = np.linalg.pinv(matrix5)

print("\n5. Матрица:")
print(matrix5)

print("Псевдообратная матрица:")
print(pseudo_inverse)
```

```python
# 6. Извлечение чётных строк матрицы 5x5
matrix6 = np.arange(1, 26).reshape(5, 5)

even_rows = matrix6[1::2]

print("\n6. Матрица:")
print(matrix6)

print("Чётные строки:")
print(even_rows)
```

```python
# 7. Массив 5x5 и кумулятивное произведение по столбцам
matrix7 = np.arange(1, 26).reshape(5, 5)

cum_product = np.cumprod(matrix7, axis=0)

print("\n7. Матрица:")
print(matrix7)

print("Кумулятивное произведение по столбцам:")
print(cum_product)
```

```python
# 8. Пересечение, объединение и разность массивов
arr8_1 = np.array([1, 2, 3, 4])
arr8_2 = np.array([3, 4, 5, 6])

intersection = np.intersect1d(arr8_1, arr8_2)
union = np.union1d(arr8_1, arr8_2)
difference = np.setdiff1d(arr8_1, arr8_2)

print("\n8.")
print("Пересечение:", intersection)
print("Объединение:", union)
print("Разность:", difference)
```

```python
# 9. Замена элементов больше 3 на их квадраты
arr9 = np.array([1, 2, 3, 4, 5])

result9 = np.where(arr9 > 3, arr9 ** 2, arr9)

print("\n9.")
print("Исходный массив:", arr9)
print("Результат:", result9)
```

```python
# 10. Сортировка массива по абсолютному значению
arr10 = np.array([-10, 5, -2, 8, -1, 3])

sorted_arr = arr10[np.argsort(np.abs(arr10))]

print("\n10.")
print("Исходный массив:", arr10)
print("По абсолютному значению:", sorted_arr)
```
